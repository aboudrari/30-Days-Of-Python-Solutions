****List Comprehension****

```python
# Basic syntax: [expression for item in iterable]
numbers = [1, 2, 3, 4, 5]
squares = [n ** 2 for n in numbers]
print(squares)  # [1, 4, 9, 16, 25]
```

***With a condition***

```python
numbers = range(1, 21)
even_numbers = [n for n in numbers if n % 2 == 0]
print(even_numbers)
```

***if...else inside a comprehension***

```python
numbers = range(1, 11)
labels = ['even' if n % 2 == 0 else 'odd' for n in numbers]
print(labels)
```

***Comprehension over strings***

```python
sentence = 'thirty days of python'
capitalized_words = [word.capitalize() for word in sentence.split()]
print(capitalized_words)
```

***Nested list comprehension***

```python
matrix = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]
flattened = [num for row in matrix for num in row]
print(flattened)
```

***Set and dict comprehension (recap)***

```python
unique_lengths = {len(word) for word in ['hi', 'hello', 'hey', 'ok']}
print(unique_lengths)

squares_dict = {n: n ** 2 for n in range(1, 6)}
print(squares_dict)
```

***Lambda Function***


```python
# syntax
# x = lambda param1, param2, param3: param1 + param2 + param3
# print(x(arg1, arg2, arg3))
```

```python
# Named function
def add_two_nums(a, b):
	return a + b

print(add_two_nums(2, 3))     # 5
# Lets change the above function to a lambda function
add_two_nums = lambda a, b: a + b
print(add_two_nums(2,3))    # 5

# Self invoking lambda function
(lambda a, b: a + b)(2,3) # 5 - need to encapsulate it in print() to see the result in the console

square = lambda x : x ** 2
print(square(3))    # 9
cube = lambda x : x ** 3
print(cube(3))    # 27

# Multiple variables
multiple_variable = lambda a, b, c: a ** 2 - 3 * b + 4 * c
print(multiple_variable(5, 5, 3)) # 22
```

***Lambda Function Inside Another Function***


```python
def power(x):
	return lambda n : x ** n

cube = power(2)(3)   # function power now need 2 arguments to run, in separate rounded brackets
print(cube)          # 8
two_power_of_five = power(2)(5) 
print(two_power_of_five)  # 32
```
