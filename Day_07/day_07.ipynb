****Tuples****

```python
# Creating a tuple
empty_tuple = ()  # or tuple()
print(empty_tuple)

fruits = ('banana', 'orange', 'mango', 'lemon')
print(fruits)
print(type(fruits))   # <class 'tuple'>
print(len(fruits))    # 4

# A tuple with a single item needs a trailing comma
single = ('banana',)
print(type(single))   # <class 'tuple'>
```

***Tuples are immutable***

```python
# Tuples cannot be modified after creation
fruits = ('banana', 'orange', 'mango', 'lemon')
try:
    fruits[0] = 'apple'
except TypeError as e:
    print('Error:', e)
```

***Accessing and slicing items***

```python
fruits = ('banana', 'orange', 'mango', 'lemon')
print(fruits[0])     # banana
print(fruits[-1])    # lemon
print(fruits[1:3])   # ('orange', 'mango')
print(fruits[:2])    # ('banana', 'orange')
```

***Unpacking a tuple***

```python
fruits = ('banana', 'orange', 'mango', 'lemon')
first, second, third, fourth = fruits
print(first, second, third, fourth)

# Using * to capture the rest
first, *rest = fruits
print(first)  # banana
print(rest)   # ['orange', 'mango', 'lemon']
```

***count() and index()***

```python
numbers = (1, 2, 2, 3, 4, 2)
print(numbers.count(2))  # 3
print(numbers.index(3))  # 3
```

***Concatenation, membership and conversion***

```python
front_end = ('HTML', 'CSS', 'JS')
back_end = ('Python', 'SQL')
full_stack = front_end + back_end
print(full_stack)

print('CSS' in front_end)   # True

# Converting between list and tuple
as_list = list(full_stack)
as_list.append('Docker')
back_to_tuple = tuple(as_list)
print(back_to_tuple)
```

***Nested tuples***

```python
nested = ((1, 2), (3, 4), ('a', 'b'))
print(nested[1])     # (3, 4)
print(nested[1][0])  # 3
```
