****Higher Order Functions****

```python
# A higher order function takes another function as an argument
# and/or returns a function.

def square(n):
    return n ** 2

def apply_twice(func, value):
    return func(func(value))

print(apply_twice(square, 2))  # square(square(2)) = 16
```

***Functions returning functions (closures)***

```python
def multiplier(factor):
    def multiply(n):
        return n * factor
    return multiply

double = multiplier(2)
triple = multiplier(3)
print(double(5))  # 10
print(triple(5))  # 15
```

***Lambda functions***

```python
square = lambda n: n ** 2
add = lambda a, b: a + b
print(square(5))   # 25
print(add(2, 3))   # 5
```

***map()***

```python
numbers = [1, 2, 3, 4, 5]
squared = list(map(lambda n: n ** 2, numbers))
print(squared)
```

***filter()***

```python
numbers = range(1, 21)
even_numbers = list(filter(lambda n: n % 2 == 0, numbers))
print(even_numbers)
```

***reduce() from functools***

```python
from functools import reduce
numbers = [1, 2, 3, 4, 5]
total = reduce(lambda acc, n: acc + n, numbers)
print(total)  # 15
```
