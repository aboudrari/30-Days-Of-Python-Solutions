****Dictionaries****

```python
# Creating a dictionary
empty_dict = {}  # or dict()
print(empty_dict)  # {}

dog = {
    'name': 'Rex',
    'legs': 4,
    'color': 'black and brown',
    'breed': 'German Shepherd'
}
print(dog)
print(type(dog))  # <class 'dict'>
print(len(dog))   # 4
```

***Accessing dictionary items***

```python
# Accessing dictionary items
dog = {'name': 'Rex', 'legs': 4, 'color': 'black and brown', 'breed': 'German Shepherd'}
print(dog['name'])   # Rex
print(dog['legs'])   # 4
print(dog.get('breed'))       # German Shepherd
print(dog.get('age', 'N/A'))  # N/A, default value if key does not exist
```

***keys(), values() and items()***

```python
# keys(), values() and items()
dog = {'name': 'Rex', 'legs': 4, 'color': 'black and brown', 'breed': 'German Shepherd'}
dog_keys = dog.keys()
print(dog_keys)    # dict_keys(['name', 'legs', 'color', 'breed'])
dog_values = dog.values()
print(dog_values)  # dict_values(['Rex', 4, 'black and brown', 'German Shepherd'])
dog_items = dog.items()
print(dog_items)   # dict_items([('name', 'Rex'), ('legs', 4), ...])
```

***Modifying a dictionary***

```python
# Modifying a dictionary
dog = {'name': 'Rex', 'legs': 4, 'color': 'black and brown', 'breed': 'German Shepherd'}
dog['name'] = 'Buddy'   # changing an existing value
print(dog)
dog['age'] = 3          # adding a new key-value pair
print(dog)
```

***update()***

```python
# update(): merges keys and values from one dictionary into another
dog = {'name': 'Buddy', 'legs': 4, 'color': 'black and brown', 'breed': 'German Shepherd', 'age': 3}
dog.update({'age': 4, 'is_vaccinated': True})
print(dog)
```

***Checking keys using in***

```python
# Checking if a key exists using in
dog = {'name': 'Buddy', 'legs': 4, 'color': 'black and brown', 'breed': 'German Shepherd', 'age': 4}
print('name' in dog)   # True
print('age' in dog)    # True
print('weight' in dog) # False
```

***pop()***

```python
# pop(): removes a key and returns its value
dog = {'name': 'Buddy', 'legs': 4, 'color': 'black and brown', 'breed': 'German Shepherd', 'age': 4}
age = dog.pop('age')
print(age)  # 4
print(dog)  # age is removed
```

***popitem()***

```python
# popitem(): removes the last inserted key-value pair
dog = {'name': 'Buddy', 'legs': 4, 'color': 'black and brown', 'breed': 'German Shepherd'}
last_item = dog.popitem()
print(last_item)  # ('breed', 'German Shepherd')
print(dog)
```

***del***

```python
# del: removes a key-value pair or the whole dictionary
dog = {'name': 'Buddy', 'legs': 4, 'color': 'black and brown'}
del dog['color']
print(dog)  # color is removed
```

***clear()***

```python
# clear(): empties the dictionary
dog = {'name': 'Buddy', 'legs': 4}
dog.clear()
print(dog)  # {}
```

***copy()***

```python
# copy(): creates a shallow copy of the dictionary
dog = {'name': 'Buddy', 'legs': 4}
dog_copy = dog.copy()
dog_copy['name'] = 'Max'
print(dog)       # original dictionary is unaffected
print(dog_copy)
```

***fromkeys()***

```python
# fromkeys(): creates a dictionary from a list of keys with the same default value
skills = ['Python', 'SQL', 'JavaScript']
skill_levels = dict.fromkeys(skills, 'beginner')
print(skill_levels)
```

***Looping through a dictionary***

```python
# Looping through a dictionary
person = {'first_name': 'Abdallah', 'last_name': 'Boudrari', 'country': 'Morocco'}
for key in person:
    print(key, ':', person[key])

for key, value in person.items():
    print(key, '->', value)
```

***Dictionary comprehension***

```python
# Dictionary comprehension
numbers = [1, 2, 3, 4, 5]
squares = {n: n ** 2 for n in numbers}
print(squares)  # {1: 1, 2: 4, 3: 9, 4: 16, 5: 25}
```
