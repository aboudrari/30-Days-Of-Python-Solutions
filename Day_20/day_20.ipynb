****Python Package Manager (pip)****

`pip` is Python's standard package manager. It installs, upgrades and removes third-party packages from the [Python Package Index (PyPI)](https://pypi.org).

````python
import subprocess
result = subprocess.run(['pip', '--version'], capture_output=True, text=True)
print(result.stdout)
````

***Common pip commands***

```bash
pip install requests            # install a package
pip install requests==2.31.0    # install a specific version
pip uninstall requests          # remove a package
pip list                        # list installed packages
pip show requests               # show details about a package
pip freeze > requirements.txt   # export installed packages
pip install -r requirements.txt # install from a requirements file
pip install --upgrade requests  # upgrade a package
```

````python
import subprocess
result = subprocess.run(['pip', 'list'], capture_output=True, text=True)
print('\n'.join(result.stdout.splitlines()[:10]))
````

***requirements.txt***

A `requirements.txt` file pins the exact packages (and versions) a project needs, so anyone can recreate the same environment with `pip install -r requirements.txt`.

```
requests==2.31.0
pandas==2.2.0
numpy==1.26.0
```
