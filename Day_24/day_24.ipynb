****Statistics****

```python
import statistics as stats

ages = [22, 25, 19, 34, 29, 22, 41, 25, 22]

print('Mean:', stats.mean(ages))
print('Median:', stats.median(ages))
print('Mode:', stats.mode(ages))
```

***Spread: variance and standard deviation***

```python
import statistics as stats

ages = [22, 25, 19, 34, 29, 22, 41, 25, 22]
print('Variance:', stats.variance(ages))
print('Std dev:', stats.stdev(ages))
```

***Min, max and range***

```python
ages = [22, 25, 19, 34, 29, 22, 41, 25, 22]
print('Min:', min(ages))
print('Max:', max(ages))
print('Range:', max(ages) - min(ages))
```

***Using numpy for the same statistics***

```python
import numpy as np

ages = np.array([22, 25, 19, 34, 29, 22, 41, 25, 22])
print('Mean:', ages.mean())
print('Std dev:', ages.std())
print('Percentile 90:', np.percentile(ages, 90))
```

***A simple frequency distribution***

```python
from collections import Counter

ages = [22, 25, 19, 34, 29, 22, 41, 25, 22]
freq = Counter(ages)
for value, count in sorted(freq.items()):
    print(value, ':', '*' * count)
```
