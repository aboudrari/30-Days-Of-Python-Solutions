****Python Web (Flask)****

[Flask](https://flask.palletsprojects.com/) is a lightweight web framework for building web applications and APIs in Python. The examples below are illustrative (a Flask app needs to run as a live server, so they are shown as code rather than executed in this notebook).

***A minimal Flask app***

```python
from flask import Flask

app = Flask(__name__)

@app.route('/')
def home():
    return 'Hello, World!'

if __name__ == '__main__':
    app.run(debug=True)
```

Running this and visiting `http://127.0.0.1:5000/` shows "Hello, World!".

***Routes with parameters***

```python
@app.route('/greet/<name>')
def greet(name):
    return f'Hello, {name}!'
```

***Rendering templates***

```python
from flask import render_template

@app.route('/profile/<username>')
def profile(username):
    return render_template('profile.html', username=username)
```

Templates live in a `templates/` folder and use Jinja2 syntax like `{{ username }}`.

***Handling forms and requests***

```python
from flask import request

@app.route('/submit', methods=['GET', 'POST'])
def submit():
    if request.method == 'POST':
        name = request.form['name']
        return f'Received: {name}'
    return '''
        <form method="post">
            <input name="name">
            <input type="submit">
        </form>
    '''
```
