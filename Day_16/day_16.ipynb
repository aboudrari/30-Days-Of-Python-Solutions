****Python Date & Time****

```python
from datetime import datetime

now = datetime.now()
print(now)
print(now.year, now.month, now.day)
print(now.hour, now.minute, now.second)
```

***date and time objects***

```python
from datetime import date, time

today = date.today()
print(today)

t = time(14, 30, 0)
print(t)
```

***Formatting dates with strftime()***

```python
from datetime import datetime

now = datetime.now()
print(now.strftime('%Y-%m-%d'))
print(now.strftime('%A, %d %B %Y'))
print(now.strftime('%H:%M:%S'))
```

***Parsing strings into dates with strptime()***

```python
from datetime import datetime

date_string = '2026-07-08'
parsed = datetime.strptime(date_string, '%Y-%m-%d')
print(parsed)
print(type(parsed))
```

***timedelta - date arithmetic***

```python
from datetime import datetime, timedelta

now = datetime.now()
one_week_later = now + timedelta(weeks=1)
print(one_week_later)

birth_date = datetime(2000, 1, 1)
age_delta = now - birth_date
print('Days alive:', age_delta.days)
```

***The time module***

```python
import time

start = time.time()
# simulate a bit of work
total = sum(range(1_000_000))
end = time.time()
print('Elapsed seconds:', round(end - start, 4))
```
