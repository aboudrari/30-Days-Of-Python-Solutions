****Modules****

```python
# Importing an entire module
import math
print(math.pi)
print(math.sqrt(16))
```

***import ... as (aliasing)***

```python
import statistics as stats
data = [1, 2, 3, 4, 5]
print(stats.mean(data))
print(stats.median(data))
```

***from ... import***

```python
from math import pi, sqrt
print(pi)
print(sqrt(25))
```

***Some common standard library modules***

```python
import random
print(random.randint(1, 10))
print(random.choice(['rock', 'paper', 'scissors']))

import string
print(string.ascii_lowercase)
print(string.digits)

import os
print(os.getcwd())
```

***dir() - listing what a module provides***

```python
import math
print([name for name in dir(math) if not name.startswith('_')][:10])
```
