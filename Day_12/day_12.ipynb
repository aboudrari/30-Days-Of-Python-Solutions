```python
# 1. Write a function which generates a six digit/character random_user_id.

import random
import string

def random_user_id():
	characters = string.ascii_lowercase + string.digits
	result = ""
	for _ in range(6):
		result += random.choice(characters)
	return result

print(random_user_id())
```

```python
# 2. Modify the previous task. Declare a function named user_id_gen_by_user. It doesn’t take any parameters but it takes two inputs using input(). 
# One of the inputs is the number of characters and the second input is the number of IDs which are supposed to be generated.

def user_id_gen_by_user():
	length = int(input("Enter number of characters: "))
	count = int(input("Enter number of IDs: "))
	characters = string.ascii_lowercase + string.digits
	
	for _ in range(count):        # repeat once per ID needed
		result = ""
		for _ in range(length):   # build ONE id, character by character
			result += random.choice(characters)
		print(result)             # print each id as it's made
		
print(user_id_gen_by_user())
```

```python
# 3. Write a function named rgb_color_gen. It will generate rgb colors (3 values ranging from 0 to 255 each).
import numpy as np

def rgb_color_gen():
	colors = []
	for _ in range (3):
		colors.append(np.random.randint(0, 256) )
	return tuple(colors)

print(rgb_color_gen())
```

```python
# 4. Write a function list_of_hexa_colors which returns any number of hexadecimal colors in an array (six hexadecimal numbers written after #. 
# Hexadecimal numeral system is made out of 16 symbols, 
# 0-9 and first 6 letters of the alphabet, a-f. Check the task 6 for output examples).5]


import random
import string

def generate_colors(type, nums):
	characters = string.digits + 'abcdef'
	if type == 'hexa':
		def generate_hex_color():
			result = "#"
			for _ in range(6):
				result += random.choice(characters)
			return result

		colors = []
		for _ in range(nums):
			colors.append(generate_hex_color())
		return colors
	
	elif type == 'rgb':
		def generate_rgb_color():
			r, g, b = rgb_color_gen()
			return f"rgb({r},{g},{b})"

		colors = []
		for _ in range(nums):
			colors.append(generate_rgb_color())
		return colors


print(generate_colors('hexa', 3))  # ['#a3e12f', '#03ed55', '#eb3d2b']
print(generate_colors('hexa', 1))  # ['#b334ef']
print(generate_colors('rgb', 3))
print(generate_colors('rgb', 1))
```

```python
# 5. Write a function list_of_rgb_colors which returns any number of RGB colors in an array.

def list_of_rgb_colors(nums):
	def generate_rgb_color():
		r, g, b = rgb_color_gen()
		return f"rgb({r},{g},{b})"

	colors = []
	for _ in range(nums):
		colors.append(generate_rgb_color())
	return colors

print(list_of_rgb_colors(3))
print(list_of_rgb_colors(1))
```
