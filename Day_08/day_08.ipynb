****Sets****

```python
# Creating a set
empty_set = set()  # note: {} creates an empty dict, not a set
print(type(empty_set))

fruits = {'banana', 'orange', 'mango', 'lemon'}
print(fruits)
print(type(fruits))   # <class 'set'>
print(len(fruits))

# Sets automatically remove duplicates
numbers = {1, 2, 2, 3, 3, 3, 4}
print(numbers)  # {1, 2, 3, 4}
```

***add(), update() and remove()***

```python
fruits = {'banana', 'orange', 'mango'}
fruits.add('lemon')
print(fruits)

fruits.update(['apple', 'kiwi'])
print(fruits)

fruits.remove('banana')   # raises KeyError if not found
print(fruits)

fruits.discard('not-there')  # does nothing if not found, no error
print(fruits)
```

***pop() and clear()***

```python
fruits = {'banana', 'orange', 'mango'}
removed = fruits.pop()  # removes an arbitrary item
print('removed:', removed)
print(fruits)

fruits.clear()
print(fruits)  # set()
```

***Set operations: union, intersection, difference***

```python
it_companies = {'Facebook', 'Google', 'Microsoft', 'Apple'}
tech_companies = {'Google', 'Apple', 'Tesla', 'SpaceX'}

print(it_companies.union(tech_companies))
print(it_companies.intersection(tech_companies))
print(it_companies.difference(tech_companies))
print(tech_companies.difference(it_companies))
print(it_companies.symmetric_difference(tech_companies))
```

***issubset(), issuperset() and isdisjoint()***

```python
a = {1, 2, 3}
b = {1, 2, 3, 4, 5}
c = {9, 10}

print(a.issubset(b))     # True
print(b.issuperset(a))   # True
print(a.isdisjoint(c))   # True, no common elements
```

***Membership and looping***

```python
fruits = {'banana', 'orange', 'mango'}
print('banana' in fruits)   # True

for fruit in fruits:
    print(fruit)
```
