****Virtual Environment****

A virtual environment is an isolated Python environment with its own installed packages, separate from the system-wide Python installation. It keeps project dependencies from clashing with each other.

````python
import subprocess
result = subprocess.run(['python', '-m', 'venv', '--help'], capture_output=True, text=True)
print(result.stdout[:600])
````

***Creating and using a virtual environment***

```bash
# Create a virtual environment named 'venv'
python -m venv venv

# Activate it (Windows PowerShell)
.\\venv\\Scripts\\Activate.ps1

# Activate it (macOS / Linux)
source venv/bin/activate

# Install packages only inside this environment
pip install requests pandas

# Leave the virtual environment
deactivate
```

***Why use one?***

- Different projects can require different (and conflicting) versions of the same package.
- Keeps your global Python installation clean.
- Makes projects reproducible: ship a `requirements.txt` and anyone can rebuild the exact environment.
- Common tools: `venv` (built into Python), `virtualenv`, `conda`, `pipenv`, `poetry`.
