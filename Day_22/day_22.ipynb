****Web Scraping****

We'll scrape [quotes.toscrape.com](http://quotes.toscrape.com), a site built specifically for scraping practice.

```python
import requests
from bs4 import BeautifulSoup

response = requests.get('http://quotes.toscrape.com')
print(response.status_code)   # 200 means success

soup = BeautifulSoup(response.text, 'html.parser')
print(soup.title.text)
```

***find() and find_all()***

```python
quotes = soup.find_all('span', class_='text')
for q in quotes[:5]:
    print(q.text)
```

***Extracting attributes***

```python
authors = soup.find_all('small', class_='author')
for a in authors[:5]:
    print(a.text)

# links have an href attribute
tags = soup.find('div', class_='tags')
print(tags.find('a')['href'] if tags else None)
```

***CSS selectors with select()***

```python
titles = soup.select('.text')
print(len(titles))
print(titles[0].get_text())
```

***Putting it together: quote + author pairs***

```python
data = []
for card in soup.find_all('div', class_='quote'):
    text = card.find('span', class_='text').get_text()
    author = card.find('small', class_='author').get_text()
    data.append((author, text))

for author, text in data[:3]:
    print(f'{author}: {text}')
```
