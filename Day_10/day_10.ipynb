****Loops****

```python
# while loop
count = 0
while count < 5:
    print(count)
    count += 1
```

***for loop***

```python
for letter in 'Python':
    print(letter)

fruits = ['banana', 'orange', 'mango']
for fruit in fruits:
    print(fruit)
```

***range()***

```python
for i in range(5):
    print(i)          # 0 1 2 3 4

for i in range(2, 10, 2):
    print(i)          # 2 4 6 8
```

***break, continue and pass***

```python
for i in range(10):
    if i == 5:
        break
    print(i)          # 0 1 2 3 4

for i in range(10):
    if i % 2 == 0:
        continue
    print(i)           # 1 3 5 7 9

for i in range(3):
    pass  # placeholder, does nothing
```

***enumerate()***

```python
fruits = ['banana', 'orange', 'mango']
for index, fruit in enumerate(fruits):
    print(index, fruit)
```

***Nested loops***

```python
for i in range(3):
    for j in range(3):
        print(f'({i}, {j})', end=' ')
    print()
```

***else clause on loops***

```python
for i in range(5):
    print(i)
else:
    print('Loop finished without break')
```

# Exercises: Day 10

```python
# 1. Use a while loop to print numbers from 0 to 10

number = 1
while number != 11:
    print(number)
    number+=1
```

```python
# 2. Use a for loop to print numbers from 0 to 10 using range()
for i in range(11):
    print(i)
```

```python
# 3. Iterate over the string 'Python' and print each letter

for chr in 'Python':
    print(chr)
```

```python
# 4. Iterate over a list of your favorite fruits and print each one with its index using enumerate()

fruits = ['apple', 'banana', 'orange', 'mango', 'strawberry', 'grape', 'watermelon', 'pineapple', 'kiwi', 'blueberry']
for index, fruit in enumerate(fruits):
    print(index , fruit)
```

```python
# 5. Print all even numbers between 1 and 100 using a for loop

for num in range (1, 100 + 1):
    if num % 2 == 0:
        print(num)
```

```python
# 6. Print the sum of all numbers from 1 to 50 using a loop
sum_ = 0
for num in range(1, 51):
    sum_ += num
print(sum_)
```

```python
# 7. Use nested loops to print a 5x5 grid of stars (*)

for i in range(5):
    row = ""
    for j in range(5):
        row += "*"
    print(row)
```

```python
# 8. Write a loop that prints numbers 1-20 but stops (break) as soon as it reaches a multiple of 13

for i in range(1, 21):
    if i % 13 == 0 :
        break
    print(i)
```

```python
# 9. Write a loop that prints only odd numbers between 1 and 20 using continue

for num in range(1,21):
    if num % 2 == 0 :
        continue
    print(num)
```

```python
# 10. Print the multiplication table (1-10) of a given number using a for loop

def multiplication_table(n):
    for i in range(1, 11):
        print(f"{n} * {i} = {n*i}")

multiplication_table(5)
```

```python
# 11. Write a program that counts down from 10 to 1 and then prints 'Liftoff!'

for num in range(10, 0, -1):
    print(num)
print("Liftoff!")
```
