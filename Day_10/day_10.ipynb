****Loops****

```python
# while loop
count = 0
while count < 5:
    print(count)
    count += 1
```

***for loop***

```python
for letter in 'Python':
    print(letter)

fruits = ['banana', 'orange', 'mango']
for fruit in fruits:
    print(fruit)
```

***range()***

```python
for i in range(5):
    print(i)          # 0 1 2 3 4

for i in range(2, 10, 2):
    print(i)          # 2 4 6 8
```

***break, continue and pass***

```python
for i in range(10):
    if i == 5:
        break
    print(i)          # 0 1 2 3 4

for i in range(10):
    if i % 2 == 0:
        continue
    print(i)           # 1 3 5 7 9

for i in range(3):
    pass  # placeholder, does nothing
```

***enumerate()***

```python
fruits = ['banana', 'orange', 'mango']
for index, fruit in enumerate(fruits):
    print(index, fruit)
```

***Nested loops***

```python
for i in range(3):
    for j in range(3):
        print(f'({i}, {j})', end=' ')
    print()
```

***else clause on loops***

```python
for i in range(5):
    print(i)
else:
    print('Loop finished without break')
```
