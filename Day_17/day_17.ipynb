***Packing and Unpacking Arguments in Python***

```python
# We use two operators:
# * for tuples
# ** for dictionaries
```

***Unpacking***

```python
def sum_of_five_nums(a, b, c, d, e):
	return a + b + c + d + e

lst = [1, 2, 3, 4, 5]
print(sum_of_five_nums(lst)) # TypeError: sum_of_five_nums() missing 4 required positional arguments: 'b', 'c', 'd', and 'e'
```

```python
def sum_of_five_nums(a, b, c, d, e):
	return a + b + c + d + e

lst = [1, 2, 3, 4, 5]
print(sum_of_five_nums(*lst))  # 15
```

```python
# We can also use unpacking in the range built-in function that expects a start and an end.
numbers = range(2, 7)  # normal call with separate arguments
print(list(numbers)) # [2, 3, 4, 5, 6]
args = [2, 7]
numbers = range(*args)  # call with arguments unpacked from a list
print(numbers)      # [2, 3, 4, 5,6]
```

```python
#A list or a tuple can also be unpacked like this:

countries = ['Finland', 'Sweden', 'Norway', 'Denmark', 'Iceland']
fin, sw, nor, *rest = countries
print(fin, sw, nor, rest)   # Finland Sweden Norway ['Denmark', 'Iceland']
numbers = [1, 2, 3, 4, 5, 6, 7]
one, *middle, last = numbers
print(one, middle, last)      #  1 [2, 3, 4, 5, 6] 7
```

```python
# Unpacking Dictionaries

def unpacking_person_info(name, country, city, age):
	return f'{name} lives in {country}, {city}. He is {age} year old.'
dct = {'name':'Asabeneh', 'country':'Finland', 'city':'Helsinki', 'age':250}
print(unpacking_person_info(**dct)) # Asabeneh lives in Finland, Helsinki. He is 250 years old.
```

***Packing***

```python
def sum_all(*args):
	s = 0
	for i in args:
		s += i
	return s
print(sum_all(1, 2, 3))             # 6
print(sum_all(1, 2, 3, 4, 5, 6, 7)) # 28
```

```python
# Packing Dictionaries

def packing_person_info(**kwargs):
	# check the type of kwargs and it is a dict type
	# print(type(kwargs))
	# Printing dictionary items
	for key in kwargs:
		print(f"{key} = {kwargs[key]}")
	return kwargs

print(packing_person_info(name="abdallah",
	  country="Morocco", city="Belfaa", age=25))
```

***Spreading in Python***


```python
# Like in JavaScript, spreading is possible in Python. Let us check it in an example below:
lst_one = [1, 2, 3]
lst_two = [4, 5, 6, 7]
lst = [0, *lst_one, *lst_two]
print(lst)          # [0, 1, 2, 3, 4, 5, 6, 7]
country_lst_one = ['Finland', 'Sweden', 'Norway']
country_lst_two = ['Denmark', 'Iceland']
nordic_countries = [*country_lst_one, *country_lst_two]
print(nordic_countries)  # ['Finland', 'Sweden', 'Norway', 'Denmark', 'Iceland']
```

***Enumerate***

```python
# If we are interested in an index of a list, we use enumerate built-in function to get the index of each item in the list.

for index, item in enumerate([20, 30, 40]):
	print(index, item)
```

```python
countries = ['Finland', 'Sweden', 'Norway', 'Denmark', 'Iceland']
for index, i in enumerate(countries):
	if i == 'Finland':
		print(f'The country {i} has been found at index {index}')
```

***Zip***

```python
# Sometimes we would like to combine lists when looping through them. See the example below:

fruits = ['banana', 'orange', 'mango', 'lemon', 'lime']                    
vegetables = ['Tomato', 'Potato', 'Cabbage','Onion', 'Carrot']
fruits_and_veges = []
for f, v in zip(fruits, vegetables):
	fruits_and_veges.append({'fruit':f, 'veg':v})

print(fruits_and_veges)
```

```python
# 1. Write a function that divides two numbers and handles ZeroDivisionError gracefully

def divid(a, b):
	try :
		result = a/b
		print(result)

	except ZeroDivisionError:
		print('zero division error occured')
divid(2,0)
```

```python
# 2. Write a function that converts user input to an integer and handles ValueError if the input is not a valid number

def handel(num):
    try :
        print(int(num))
    except ValueError:
        print("The input must be integer")
    
handel(4)
handel("jj")
```

```python
# 3. Write a function that accesses a list index and handles IndexError if the index is out of range

def access_index(lst, index):
    try:
        print(lst[index])
    except IndexError:
        print("Index out of range")

access_index([1, 2, 3], 1)
access_index([1, 2, 3], 10)
```

```python
# 4. Write a function that accesses a dictionary key and handles KeyError if the key doesn't exist

def get_dict_value(data, key):
    try:
        print(data[key])
    except KeyError:
        print("The key not found")

get_dict_value({'a': 1, 'b': 2}, 'a')
get_dict_value({'a': 1, 'b': 2}, 'z')
```

```python
# 5. Create a custom exception called NegativeNumberError and raise it when a function receives a negative number

class NegativeNumberError(Exception):
    pass

def check_positive(num):
    if num < 0:
        raise NegativeNumberError("Number cannot be negative")
    print(num)

try:
    check_positive(-5)
except NegativeNumberError as e:
    print(e)
```

```python
# 6. Write a try/except/else/finally block that opens a calculation, only proceeds on success, and always prints a closing message

def calculate(a, b):
    try:
        result = a / b
    except ZeroDivisionError:
        print("Cannot divide by zero")
    else:
        print("Calculation succeeded:", result)
    finally:
        print("Closing message: done processing")

calculate(10, 2)
print(" ")
calculate(10, 0)
```

```python
# 7. Write a function validate_age(age) that raises a ValueError for ages outside 0-120, and test it with a few values

def validate_age(age):
    if age < 0 or age > 120:
        raise ValueError("Age must be between 0 and 120")
    print(f"Valid age: {age}")

for test_value in [25, -5, 150, 0, 120]:
    try:
        validate_age(test_value)
    except ValueError as e:
        print(e)
```
