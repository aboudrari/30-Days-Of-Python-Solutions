****Pandas****

```python
import pandas as pd

# A Series is a one-dimensional labeled array
ages = pd.Series([22, 25, 19, 34], name='age')
print(ages)
```

***Creating a DataFrame***

```python
import pandas as pd

data = {
    'name': ['Alice', 'Bob', 'Charlie', 'Diana'],
    'age': [25, 30, 35, 28],
    'city': ['Casablanca', 'Rabat', 'Marrakesh', 'Fes']
}
df = pd.DataFrame(data)
print(df)
```

***Inspecting a DataFrame***

```python
print(df.head(2))
print(df.shape)     # (rows, columns)
print(df.columns.tolist())
print(df.describe())
```

***Selecting columns and rows***

```python
print(df['name'])          # a single column (Series)
print(df[['name', 'age']]) # multiple columns

print(df.loc[0])           # row by label
print(df.iloc[1])          # row by position
```

***Filtering rows***

```python
adults_over_28 = df[df['age'] > 28]
print(adults_over_28)
```

***Sorting and adding columns***

```python
df['is_adult'] = df['age'] >= 18
print(df.sort_values('age', ascending=False))
```
