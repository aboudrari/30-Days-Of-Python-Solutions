****Python with MongoDB****

MongoDB is a NoSQL, document-oriented database. Python talks to it through the `pymongo` driver. The snippets below are illustrative — they need a running MongoDB instance (local or [MongoDB Atlas](https://www.mongodb.com/atlas)) so they are shown as code rather than executed here.

***Connecting to MongoDB***

```python
from pymongo import MongoClient

client = MongoClient('mongodb://localhost:27017/')
db = client['thirty_days_of_python']
students = db['students']
```

***Inserting documents***

```python
student = {
    'name': 'Abdallah',
    'country': 'Morocco',
    'skills': ['Python', 'JavaScript']
}
result = students.insert_one(student)
print(result.inserted_id)

many_students = [
    {'name': 'Alice', 'country': 'USA'},
    {'name': 'Bob', 'country': 'Canada'}
]
students.insert_many(many_students)
```

***Querying documents***

```python
# Find one
print(students.find_one({'name': 'Abdallah'}))

# Find many
for s in students.find({'country': 'USA'}):
    print(s)
```

***Updating and deleting documents***

```python
students.update_one({'name': 'Abdallah'}, {'$set': {'country': 'Finland'}})
students.delete_one({'name': 'Bob'})
students.delete_many({'country': 'USA'})
```
