****Python Type Errors****

```python
# type() tells you the exact type of an object
print(type(10))        # <class 'int'>
print(type(10.5))      # <class 'float'>
print(type('hello'))   # <class 'str'>
print(type([1, 2]))    # <class 'list'>
```

***isinstance() - the safer way to check types***

```python
value = 10
print(isinstance(value, int))          # True
print(isinstance(value, (int, float))) # True, checks against a tuple of types
```

***Common causes of TypeError***

```python
try:
    result = '5' + 5
except TypeError as e:
    print('TypeError:', e)

try:
    result = len(10)
except TypeError as e:
    print('TypeError:', e)
```

***Converting types to avoid TypeErrors***

```python
age = '25'
print('Next year I will be', int(age) + 1)

price = 19.99
print('Price: $' + str(price))
```

***Type hints (annotations)***

```python
def add(a: int, b: int) -> int:
    return a + b

print(add(2, 3))
# Type hints are not enforced at runtime, but tools like mypy can check them
print(add.__annotations__)
```
