****Conditionals****

```python
# if statement
age = 20
if age >= 18:
	print('You are an adult')
```

***if / elif / else***

```python
weight = 75

if weight < 18.5:
	print('Underweight')
elif weight < 25:
	print('Normal weight')
elif weight < 30:
	print('Overweight')
else:
	print('Obese')
```

***Comparison and logical operators***

```python
a = 10
b = 20
print(a == b, a != b, a < b, a > b, a <= b, a >= b)

is_sunny = True
is_warm = False
print(is_sunny and is_warm)  # False
print(is_sunny or is_warm)   # True
print(not is_sunny)          # False
```

***Nested conditionals***

```python
num = 15
if num > 0:
	if num % 2 == 0:
		print('Positive even number')
	else:
		print('Positive odd number')
else:
	print('Not a positive number')
```

***Ternary (conditional) expression***

```python
num = 5
result = 'even' if num % 2 == 0 else 'odd'
print(result)
```

```python
# 1. Write a program that checks if a number is positive, negative or zero

def check_num(num):
	if num > 0:
		return "positive"
	elif num < 0:
		return "negative"
	else:
		return "zero"

print(check_num(-22))
```

```python
# 2. Write a program that checks if a number is even or odd

def ev_or_odd(num):
	if num % 2 == 0:
		print("even")
	else:
		print("odd")

ev_or_odd(13)
```

```python
# 3. Write a program that checks whether a given year is a leap year or not

def is_leap_year(year):
    if year % 4 == 0 and (year % 100 != 0 or year % 400 == 0):
	   return True
    else:
	   return False
    
print(is_leap_year(365))
print(is_leap_year(366))
```

```python
# 4. Write a program that takes a person's age and prints 'child', 'teenager', 'adult' or 'senior' accordingly

# child → under 13
# teenager → 13 to 19
# adult → 20 to 64
# senior → 65 and up

def ages(age):
    if age < 13:
        print("child")
    elif age <= 19:
        print("teenager")
    elif age <= 64:
        print("adult")
    else:
        print("senior")
          
print(ages(23))
```
