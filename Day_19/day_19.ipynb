****File Handling****

```python
# Writing to a file
with open('sample.txt', 'w') as f:
    f.write('Thirty Days Of Python\n')
    f.write('Day 19: File Handling\n')

print('File written')
```

***Reading a file***

```python
with open('sample.txt', 'r') as f:
    content = f.read()
print(content)
```

***Reading line by line***

```python
with open('sample.txt', 'r') as f:
    for line in f:
        print(line.strip())

with open('sample.txt', 'r') as f:
    lines = f.readlines()
print(lines)
```

***Appending to a file***

```python
with open('sample.txt', 'a') as f:
    f.write('Appended line\n')

with open('sample.txt', 'r') as f:
    print(f.read())
```

***Why use the `with` statement***

```python
# `with` automatically closes the file, even if an error occurs
with open('sample.txt', 'r') as f:
    pass
print('File closed automatically:', f.closed)  # True
```

***Checking file existence with os.path***

```python
import os

print(os.path.exists('sample.txt'))
print(os.path.exists('does_not_exist.txt'))
```
